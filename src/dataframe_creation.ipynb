```python
# Import Pandas for creating and analyzing DataFrames.
import pandas as pd
# Import NumPy for working with numerical arrays.
import numpy as np
```

```python
# Store JavaBeans coffee sales for 2021 and 2022.
# Each row represents a year, and each column represents a coffee type.
# Create a DataFrame with descriptive row and column labels
coffee_sales = pd.DataFrame(
    np.array([
        [5628, 12426, 852, 8522],
        [6239, 25384, 756, 7561]
    ]),
    index=["2021 Sales", "2022 Sales"],
    columns=["Americano", "Latte", "Espresso", "Cappuccino"]
)

coffee_sales


# Display the DataFrame to compare sales across both years.
```

```python

# Display the total sales for each coffee type.
total_sales = coffee_sales.sum()

total_sales
```

```python
# Calculate year-over-year growth for each coffee type.
# Subtract 2021 sales from 2022 sales, divide by 2021 sales,
# and multiply by 100 to convert the result into a percentage.

yoy_growth = ((coffee_sales.loc["2022 Sales"] - coffee_sales.loc["2021 Sales"])
              / coffee_sales.loc["2021 Sales"]) * 100

yoy_growth

# Display growth percentages rounded to two decimal places.
```
